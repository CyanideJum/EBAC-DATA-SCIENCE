```python
print('Insira o primeiro número: ')
numero1 = float(input())
print('Insira o segundo número: ')
numero2 = float(input())

print('Qual o operador matemático deseja usar? ')
operador = input()

if operador == '+':
    print(numero1 + numero2)
elif operador == '-':
    print(numero1 - numero2)
elif operador == '*':
    print(numero1 * numero2)
elif operador == '/':
    print(numero1 / numero2)
```
